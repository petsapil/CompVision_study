In this homework you are going to implement the **Floyd-Steinberg dithering** algorithm. Dithering, in general, means that we are adding noise to the signal (in our case digital image) in order to perceive it better. In other words, by adding the noise the objective quality will be worse but the subjective quality will be better (i.e. the image will "look" better).

The details of FS dithering can be found in this [wiki](https://en.wikipedia.org/wiki/Floyd%E2%80%93Steinberg_dithering) page. In order to implement the dithering, we will implement the following steps:
* Define colour palette
* Quantize the image to obtain the baseline and compute the average quantization error
* Implement FS dithering and compute the average quantization error

```python
import cv2
import math
import numpy as np
from matplotlib import pyplot as plt
plt.rcParams['figure.figsize'] = [15, 10]
```

Let's load the image.

```python
# Load image
img = cv2.imread('../data/kodim23.png')
# Convert it to RGB
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
# Plot it
plt.imshow(img)
```

Let's start with gray tones first.

```python
# Black, dark gray, light gray, white
colors = np.array([[0, 0, 0],
                   [64, 64, 64],
                   [192, 192, 192],
                   [255, 255, 255]])
```

Both parts of the homework need the same operation: given a pixel, find the closest colour of the
palette. Let's write it once as a small function and reuse it.

```python
def closest_color(pixel, colors):
    # Compute the distance from the pixel to every colour of the palette
    # Hint: colors - pixel gives you all the differences at once
    distances = np.array(colors) - pixel
    distances = np.linalg.norm(distances, axis = 1)
    # Return the colour of the palette with the smallest distance
    return colors[np.argmin(distances)]
```

Using the colour palette, let's quantize the original image. This is our baseline: every pixel is
replaced by the closest palette colour, independently of its neighbours.

```python
# Cast the image to float (the quantization error can be negative and larger than 255)
def quant(img, colors):
    img = img.astype(np.float32)
    
    # Prepare for quantization
    rows, cols, channels = img.shape
    quantized = np.zeros_like(img)
    
    # Apply quantization
    for r in range(rows):
        for c in range(cols):
            # Extract the original pixel value
            pixel = img[r,c]
    
            # Find the closest colour from the palette
            new_pixel = closest_color(pixel, colors)
    
            # Apply quantization
            quantized[r, c, :] = new_pixel
    return quantized
quantized = quant(img, colors)
```

```python
# Show quantized image (don't forget to cast back to uint8)
plt.imshow(quantized.astype(np.uint8))
```

To be able to compare the two approaches we need a number. Let's use the **average quantization error**,
i.e. the mean absolute difference between the original and the quantized image, averaged over all pixels
and all three channels.

```python
# Compute average quantization error
avg_quant_error = np.mean(np.abs(img - quantized))
print(f"avg_quant_error = {avg_quant_error}")
```

#### Floyd-Steinberg Dithering

We are now going to implement the FS dithering and compare it to the optimally quantized image we have
calculated above.

The idea is a **feedback loop**: after quantizing a pixel we know exactly how much we got it wrong,

$$ \varepsilon = p_{orig} - p_{quant} $$

and we push that error onto the neighbours that have not been processed yet. If a pixel was made too
dark, its neighbours are made brighter on purpose, so that the error averages out to zero over a small
area. The error is distributed according to the FS diffusion matrix (`*` marks the current pixel, the
image is scanned left to right, top to bottom)

$$ \frac{1}{16}
\begin{bmatrix}
 - & * & 7 \\
 3 & 5 & 1
\end{bmatrix} $$

so the pixel to the right receives 7/16 of the error, the one below-left 3/16, the one below 5/16 and
the one below-right 1/16.

Two practical notes:
* The diffused error has to be accumulated in a **copy** of the image (`img_tmp`). Each pixel must be
  quantized using the value that already contains the error of its neighbours, not the original value.
* We skip the first and the last row and column so that we do not have to handle the borders, which is
  why the loop starts at 1.

```python
# Make a temporal copy of the original image, we will need it for error diffusion
def dither(img, colors):
    rows, cols, channels = img.shape
    img_tmp = np.copy(img.astype(np.float32))
    dithering = np.zeros_like(img.astype(np.float32))
    
    for r in range(1, rows-1):
        for c in range(1, cols-1):
            # Extract the pixel value, including the error diffused from the neighbours
            pixel = img_tmp[r, c, :]
    
            # Find the closest colour from the palette
            new_pixel = closest_color(pixel, colors)
    
            # Compute quantization error
            quant_error = pixel - new_pixel
    
            # Diffuse the quantization error according to the FS diffusion matrix
            # Note: You need more than one line of code here
            if c + 1 < img.shape[1]:
                img_tmp[r, c+1, :] += quant_error * 7/16
            if r + 1 < img.shape[0]:
                if c - 1 >= 0:
                    img_tmp[r+1, c-1, :] += quant_error * 3/16
                img_tmp[r+1, c, :] += quant_error * 5/16
                if c + 1 < img.shape[1]:
                    img_tmp[r+1, c+1, :] += quant_error * 1/16
    
            # Apply dithering
            dithering[r, c, :] = new_pixel
    return dithering
dithering = dither(img, colors)
```

```python
# Show quantized image (don't forget to cast back to uint8)
plt.subplot(121), plt.imshow(quantized.astype(np.uint8))   # optimally quantized
plt.subplot(122), plt.imshow(dithering.astype(np.uint8))   # dithering
```

```python
# Compute average quantization error for dithered image
avg_dith_error = np.mean(np.abs(img - dithering))
print(f"avg_dith_error = {avg_dith_error}")
```

### Questions
* Which image has higher quantization error? Optimally quantized or dithered?
* Calculated average error shows that the dithered image accumulated higher quantization error and mathematically straight forward quantized image is closer to the original than dithered one

```python
# Black, white
colors_bl_wt = np.array([[0, 0, 0],
                   [255, 255, 255]])

quantized_2 = quantized = quant(img, colors_bl_wt)
dither_2 = dither(img, colors_bl_wt)

plt.subplot(121), plt.imshow(quantized_2.astype(np.uint8))
plt.subplot(122), plt.imshow(dither_2.astype(np.uint8))
```

### Bonus Points

Repeat the homework using a different image palette. For instance, you can use an optimal colour
palette that we can calculate via k-means algorithm. The following snippet of code will give you the 16
optimal colours for your original image.

```python
from sklearn.cluster import KMeans
kmeans = KMeans(n_clusters=16).fit(np.reshape(img, (-1, 3)))
colors_optimal_16 = kmeans.cluster_centers_
quantized_optimal_16 = quant(img, colors_optimal_16)
dither_optimal_16 = dither(img, colors_optimal_16)

kmeans = KMeans(n_clusters=32).fit(np.reshape(img, (-1, 3)))
colors_optimal_32 = kmeans.cluster_centers_
quantized_optimal_32 = quant(img, colors_optimal_32)
dither_optimal_32 = dither(img, colors_optimal_32)

kmeans = KMeans(n_clusters=256).fit(np.reshape(img, (-1, 3)))
colors_optimal_256 = kmeans.cluster_centers_
quantized_optimal_256 = quant(img, colors_optimal_256)
dither_optimal_256 = dither(img, colors_optimal_256)

plt.subplot(421), plt.imshow(np.clip(img, 0, 255).astype(np.uint8))
plt.subplot(423), plt.imshow(np.clip(quantized_optimal_16, 0, 255).astype(np.uint8))
plt.subplot(424), plt.imshow(np.clip(dither_optimal_16, 0, 255).astype(np.uint8))
plt.subplot(425), plt.imshow(np.clip(quantized_optimal_32, 0, 255).astype(np.uint8))
plt.subplot(426), plt.imshow(np.clip(dither_optimal_32, 0, 255).astype(np.uint8))
plt.subplot(427), plt.imshow(np.clip(quantized_optimal_256, 0, 255).astype(np.uint8))
plt.subplot(428), plt.imshow(np.clip(dither_optimal_256, 0, 255).astype(np.uint8))
```
